```python
from torchvision.datasets import VOCDetection

train_dataset = VOCDetection(
    root="/content/data",
    year="2012",
    image_set="train",
    download=True
)

val_dataset = VOCDetection(
    root="/content/data",
    year="2012",
    image_set="val",
    download=True
)

print("Train images:", len(train_dataset))
print("Validation images:", len(val_dataset))
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# ავიღოთ ერთი სურათი
image, annotation = train_dataset[0]

print("Image size:", image.size)
print("Image format:", image.mode)

# ვნახოთ ობიექტების ანოტაციები
objects = annotation["annotation"]["object"]

if isinstance(objects, dict):
    objects = [objects]

print("\nObjects in image:")
for obj in objects:
    print(
        "Class:", obj["name"],
        "| Bounding box:", obj["bndbox"]
    )

# გამოვსახოთ სურათი
fig, ax = plt.subplots(figsize=(10, 7))
ax.imshow(image)

for obj in objects:
    box = obj["bndbox"]

    xmin = int(box["xmin"])
    ymin = int(box["ymin"])
    xmax = int(box["xmax"])
    ymax = int(box["ymax"])

    rect = patches.Rectangle(
        (xmin, ymin),
        xmax - xmin,
        ymax - ymin,
        linewidth=2,
        edgecolor="red",
        facecolor="none"
    )

    ax.add_patch(rect)
    ax.text(
        xmin, max(0, ymin - 5),
        obj["name"],
        color="white",
        backgroundcolor="red"
    )

plt.axis("off")
plt.show()
```

```python
from collections import Counter

class_counts = Counter()

for _, annotation in train_dataset:
    objects = annotation["annotation"].get("object", [])

    if isinstance(objects, dict):
        objects = [objects]

    for obj in objects:
        class_counts[obj["name"]] += 1

print("Number of classes:", len(class_counts))
print("\nObjects per class:")

for class_name, count in class_counts.most_common():
    print(f"{class_name}: {count}")
```

```python
from pathlib import Path
from collections import Counter
import shutil
import yaml

# YOLO dataset-ის საქაღალდე
BASE_DIR = Path("/content/voc_yolo")

for split in ["train", "val"]:
    (BASE_DIR / "images" / split).mkdir(
        parents=True, exist_ok=True
    )
    (BASE_DIR / "labels" / split).mkdir(
        parents=True, exist_ok=True
    )

# Pascal VOC-ის 20 კლასი
CLASS_NAMES = [
    "aeroplane", "bicycle", "bird", "boat", "bottle",
    "bus", "car", "cat", "chair", "cow", "diningtable",
    "dog", "horse", "motorbike", "person", "pottedplant",
    "sheep", "sofa", "train", "tvmonitor"
]

CLASS_TO_ID = {
    name: idx for idx, name in enumerate(CLASS_NAMES)
}


def convert_dataset(dataset, split):
    image_dir = BASE_DIR / "images" / split
    label_dir = BASE_DIR / "labels" / split

    total_objects = 0
    skipped_objects = Counter()

    for idx in range(len(dataset)):
        image, annotation = dataset[idx]

        # სურათის შენახვა
        image_path = image_dir / f"{idx:06d}.jpg"
        image.save(image_path)

        width, height = image.size
        objects = annotation["annotation"].get("object", [])

        if isinstance(objects, dict):
            objects = [objects]

        label_lines = []

        for obj in objects:
            class_name = obj["name"]

            if class_name not in CLASS_TO_ID:
                skipped_objects[class_name] += 1
                continue

            box = obj["bndbox"]

            xmin = float(box["xmin"])
            ymin = float(box["ymin"])
            xmax = float(box["xmax"])
            ymax = float(box["ymax"])

            # კოორდინატების შემოწმება
            if xmax <= xmin or ymax <= ymin:
                skipped_objects["invalid_box"] += 1
                continue

            # პიქსელებიდან ნორმალიზებულ კოორდინატებზე გადასვლა
            x_center = ((xmin + xmax) / 2) / width
            y_center = ((ymin + ymax) / 2) / height
            box_width = (xmax - xmin) / width
            box_height = (ymax - ymin) / height

            class_id = CLASS_TO_ID[class_name]

            label_lines.append(
                f"{class_id} {x_center:.6f} {y_center:.6f} "
                f"{box_width:.6f} {box_height:.6f}"
            )
            total_objects += 1

        label_path = label_dir / f"{idx:06d}.txt"
        label_path.write_text(
            "\n".join(label_lines),
            encoding="utf-8"
        )

    print(f"{split}: {len(dataset)} images processed")
    print(f"{split}: {total_objects} objects converted")
    print("Skipped:", dict(skipped_objects))


convert_dataset(train_dataset, "train")
convert_dataset(val_dataset, "val")

# YOLO-ს dataset configuration
data_config = {
    "path": str(BASE_DIR),
    "train": "images/train",
    "val": "images/val",
    "names": {idx: name for idx, name in enumerate(CLASS_NAMES)}
}

with open(BASE_DIR / "data.yaml", "w", encoding="utf-8") as f:
    yaml.safe_dump(
        data_config,
        f,
        sort_keys=False,
        allow_unicode=True
    )

print("\nDataset conversion complete!")
print("Configuration:", BASE_DIR / "data.yaml")
```

```python
from pathlib import Path
from collections import Counter

BASE_DIR = Path("/content/voc_yolo")

for split in ["train", "val"]:
    image_dir = BASE_DIR / "images" / split
    label_dir = BASE_DIR / "labels" / split

    images = list(image_dir.glob("*.jpg"))
    labels = list(label_dir.glob("*.txt"))

    errors = []
    class_counts = Counter()
    total_boxes = 0

    for label_file in labels:
        for line_number, line in enumerate(
            label_file.read_text().splitlines(), start=1
        ):
            parts = line.split()

            if len(parts) != 5:
                errors.append(f"{label_file.name}: wrong format")
                continue

            class_id, x, y, w, h = map(float, parts)

            if not class_id.is_integer() or not (0 <= class_id < 20):
                errors.append(f"{label_file.name}: invalid class ID")
                continue

            if not (
                0 <= x <= 1 and 0 <= y <= 1
                and 0 < w <= 1 and 0 < h <= 1
            ):
                errors.append(f"{label_file.name}: invalid coordinates")
                continue

            class_counts[int(class_id)] += 1
            total_boxes += 1

    print(f"\n--- {split.upper()} ---")
    print("Images:", len(images))
    print("Label files:", len(labels))
    print("Bounding boxes:", total_boxes)
    print("Invalid annotations:", len(errors))
    print("Classes represented:", len(class_counts))

    if errors:
        print("Examples:", errors[:5])
```

```python
from ultralytics import YOLO
from pathlib import Path

# წინასწარ გაწვრთნილი YOLO მოდელი
model = YOLO("yolo11n.pt")

# პირველი ექსპერიმენტი — baseline
results = model.train(
    data="/content/voc_yolo/data.yaml",
    epochs=10,
    imgsz=640,
    batch=16,
    device=0,
    workers=2,
    project="/content/runs",
    name="baseline",
    seed=42,
    plots=True
)

print("Training finished!")
print("Results saved to:", results.save_dir)
```

```python
from ultralytics import YOLO
from IPython.display import display
from PIL import Image
from pathlib import Path

# საუკეთესო გაწვრთნილი მოდელის ჩატვირთვა
model = YOLO("/content/runs/baseline/weights/best.pt")

# ავირჩიოთ validation-დან სურათი
image_path = "/content/voc_yolo/images/val/000000.jpg"

# ობიექტების ამოცნობა
results = model.predict(
    source=image_path,
    conf=0.25,
    save=True,
    project="/content/runs",
    name="predictions"
)

# შედეგის ჩვენება
annotated_image = results[0].plot()
display(Image.fromarray(annotated_image[:, :, ::-1]))

# აღმოჩენილი ობიექტების ჩამონათვალი
for box in results[0].boxes:
    class_id = int(box.cls[0])
    confidence = float(box.conf[0])

    print(
        f"Class: {model.names[class_id]}, "
        f"Confidence: {confidence:.2%}"
    )
```

```python
from pathlib import Path
import shutil
import zipfile

project = Path("/content/voc_cv_project")
project.mkdir(exist_ok=True)

# Copy trained model into the project
shutil.copy2(
    "/content/drive/MyDrive/best.pt",
    project / "best.pt"
)

files = {
    "requirements.txt": """ultralytics
streamlit
pillow
""",

    "app.py": '''import streamlit as st
from ultralytics import YOLO
from PIL import Image

st.set_page_config(page_title="VOC Object Detector", layout="wide")
st.title("Computer Vision: Object Detection")
st.write("YOLO11n fine-tuned on Pascal VOC 2012.")

@st.cache_resource
def load_model():
    return YOLO("best.pt")

model = load_model()
uploaded = st.file_uploader(
    "Upload an image",
    type=["jpg", "jpeg", "png"]
)

confidence = st.slider(
    "Confidence threshold",
    min_value=0.10,
    max_value=0.90,
    value=0.25,
    step=0.05
)

if uploaded is not None:
    image = Image.open(uploaded).convert("RGB")
    results = model.predict(
        source=image,
        conf=confidence,
        verbose=False
    )
    annotated = results[0].plot()
    annotated = annotated[:, :, ::-1]

    left, right = st.columns(2)
    with left:
        st.subheader("Original image")
        st.image(image, use_container_width=True)
    with right:
        st.subheader("Predictions")
        st.image(annotated, use_container_width=True)

    st.subheader("Detected objects")
    if len(results[0].boxes) == 0:
        st.write("No objects detected at this confidence threshold.")
    else:
        for box in results[0].boxes:
            class_id = int(box.cls[0])
            score = float(box.conf[0])
            st.write(f"{model.names[class_id]} — {score:.1%}")
''',

    "README.md": """# Pascal VOC Object Detection with YOLO11n

## Project overview
This project fine-tunes a pretrained YOLO11n object detector on Pascal VOC 2012.
It detects and localizes objects using bounding boxes for 20 VOC classes.

## Dataset
Pascal VOC 2012 detection dataset.
- Training split: 5,717 images
- Validation split: 5,823 images
- Training annotations converted from Pascal VOC XML to YOLO format
- 20 object classes

The validation split was used to evaluate the trained model. It is not a separate
held-out test set.

## Model
- Architecture: YOLO11n
- Initialization: pretrained `yolo11n.pt`
- Training: 10 epochs, image size 640, batch size 16
- Training seed: 42
- Hardware: Google Colab Tesla T4 GPU

## Baseline validation metrics
- Precision: 0.765
- Recall: 0.659
- mAP@50: 0.727
- mAP@50–95: 0.535

## Image-size experiment
The same trained weights were evaluated at 512 pixels without retraining.
- Precision: 0.7772
- Recall: 0.6414
- mAP@50: 0.7183
- mAP@50–95: 0.5300
- Reported inference time: 2.1 ms/image at 512 px, versus 3.0 ms/image at 640 px
  in the baseline validation log on a Tesla T4. These timings are hardware-dependent.

## Run the demo
1. Install Python dependencies: `pip install -r requirements.txt`
2. Ensure `best.pt` is in the same folder as `app.py`.
3. Run: `streamlit run app.py`
4. Upload a JPG or PNG image and adjust the confidence threshold.

## Limitations
Performance varies by class, object size, image quality, and scene complexity.
The validation split is not an independent test set. The 512-pixel experiment
shows a speed/accuracy trade-off, not a universal improvement.

## Reproducibility
The trained weights are included as `best.pt`. For retraining, prepare Pascal VOC
2012 annotations in YOLO format and use the Ultralytics YOLO training API.
Exact package versions and the dataset preparation code should be recorded for
a fully reproducible environment.
""",

    "technical_report.md": """# Technical Report: Pascal VOC Object Detection

## 1. Problem
Build a computer vision system that recognizes and localizes objects in images
using bounding boxes.

## 2. Data preparation
Pascal VOC 2012 provides XML annotations with object class names and bounding
box coordinates. These annotations were converted into YOLO format:
class_id, normalized_center_x, normalized_center_y, normalized_width,
normalized_height.

The training split contained 5,717 images and 15,774 object annotations.
The validation split contained 5,823 images and 15,787 object annotations.
Data checks found no invalid converted annotations. All 20 classes were represented.
Class frequencies were imbalanced.

## 3. Model and training
A pretrained YOLO11n detector was fine-tuned for 10 epochs using image size
640, batch size 16, and random seed 42 on a Tesla T4 GPU.

## 4. Baseline results
Precision = 0.765; Recall = 0.659; mAP@50 = 0.727; mAP@50–95 = 0.535.

Precision measures how many predicted detections are correct. Recall measures
how many ground-truth objects are detected. mAP@50 averages class-level
average precision at IoU threshold 0.50. mAP@50–95 averages over IoU thresholds
from 0.50 to 0.95 and penalizes less accurate box localization more strongly.

## 5. Experiment: image size
The same trained weights were evaluated at 512 px without retraining.

| Metric | 640 px baseline | 512 px |
|---|---:|---:|
| Precision | 0.765 | 0.7772 |
| Recall | 0.659 | 0.6414 |
| mAP@50 | 0.727 | 0.7183 |
| mAP@50–95 | 0.535 | 0.5300 |
| Inference time/image | 3.0 ms | 2.1 ms |

The smaller input size was faster in the reported validation run, while mAP
decreased slightly. Precision increased but recall decreased. This is a
speed/accuracy trade-off; results depend on hardware and evaluation conditions.

## 6. Strengths and limitations
YOLO11n is lightweight and supports fast inference. The system can detect
multiple VOC classes and display bounding boxes on uploaded images.
Limitations include class imbalance, missed or incorrect detections, sensitivity
to small objects and image conditions, and the lack of a separate held-out test set.

## 7. Future improvements
Use a separate test split, inspect per-class errors, test confidence thresholds,
compare image sizes and model variants, and evaluate performance on images
outside Pascal VOC.

## 8. AI tools disclosure
AI assistance was used to help draft code, documentation, and explanations.
The project owner should review, understand, and verify all code and results
before submission.
"""
}

for name, content in files.items():
    (project / name).write_text(content, encoding="utf-8")

zip_path = "/content/voc_cv_project.zip"
with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as z:
    for path in project.rglob("*"):
        if path.is_file():
            z.write(path, path.relative_to(project))

print("Project created:", project)
print("ZIP created:", zip_path)
print("Files:", [p.name for p in project.iterdir()])
```
